## Image 1

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load the CSV data
df_dish_pickups_loaded = pd.read_csv('dish_pickups.csv')

# Plotting dish pickups over episodes for each layout
plt.figure(figsize=(12, 6))
episodes = df_dish_pickups_loaded['episode']
rewards = df_dish_pickups_loaded['asymmetric_advantages']
rewards_1 = df_dish_pickups_loaded['forced_coordination']
rewards_2 = df_dish_pickups_loaded['cramped_room']

plt.plot(episodes, rewards, color='green', label="asymmetric_advantages")
plt.plot(episodes, rewards_2, color='blue', label="cramped_room")
plt.plot(episodes, rewards_1, color='red', label="forced_coordination")

# Add titles and labels
plt.title("Dish Pickups Per 1000 Episodes")
plt.xlabel("Episode")
plt.ylabel("Number of Dish Pickups")
plt.legend(title="Layout")

# Set y-axis ticks to whole numbers
plt.yticks([5, 10, 15, 20, 25])

plt.grid(False)
plt.show()
```

## Image 3

```python
import pandas as pd

df = pd.read_csv('Average_delivered_soups.csv')

df.head()
```

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Define the threshold value (or use an actual column from the DataFrame if it exists)
threshold_value = df['threshold'].iloc[0]  # Assuming threshold is a constant column

# Plot Reward per Episode
plt.figure(figsize=(12, 6))
episodes = df['episode']
rewards = df['asymetric_advantages']
rewards_1 = df['forced_coordination']
rewards_2 = df['cramped_room']

plt.plot(episodes, rewards, color='green', label="asymmetric_advantages")
plt.plot(episodes, rewards_2, color='blue', label="cramped_room")
plt.plot(episodes, rewards_1, color='red', label="forced_coordination")

# Add the threshold as a black dotted line
plt.axhline(y=threshold_value, color='black', linestyle='--', label="threshold")

# Add titles and labels
plt.title("Average Delivered Soups Per 1000 Episodes")
plt.xlabel("Episode")
plt.ylabel("Soups Delivered")
plt.legend()
plt.grid(False)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load the dataset
data = pd.read_csv('synthetic_delivered_soups.csv')

# Extract data for plotting
layouts = data['layout']
means = data['mean']
std_devs = data['std']

# Plot the data
plt.figure(figsize=(8, 5))
plt.errorbar(layouts, means, yerr=std_devs, fmt='o', color='blue', ecolor='blue', capsize=5, label='mean ± std')
plt.axhline(y=7, color='black', linestyle='--', label='threshold')

# Add labels, title, and legend
plt.title("Delivered Soups Per 1000 Episodes (Evaluation)")
plt.ylabel("Average of Delivered Soup per Episode")
plt.ylim(0, 12)  # Set y-axis from 0 to 10
plt.xticks(rotation=0)
plt.legend()

# Show the plot
plt.show()
```
